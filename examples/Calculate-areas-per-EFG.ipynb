```r
raster_file <- here::here(data_dir, "GET_NVIS_Pre1750_100m.tif")
r0 <- rast(raster_file)
raster_file <- here::here(data_dir, "GET_NVIS_Extant_100m.tif")
r1 <- rast(raster_file)
```

# Calcular tabla de frecuencias

```r
frequencies_pre <- freq(r0)
```

```r
frequencies_post <- freq(r1)
```

```r
total_pre_count <- sum(frequencies_pre$count)
total_post_count <- sum(frequencies_post$count)
```

```r
total_pre_count/total_post_count
```

```r

efg_freq_tables <- frequencies_pre |> 
    left_join(frequencies_post, by='value') |>
    transmute(value,
              `Proportion Pre 1750`=set_units(count.x *100/total_pre_count, "%"),
              `Proportion Extant`=set_units(count.y * 100/total_post_count, "%")) 
```

# Filter for selected functional groups

```r
slc <- c(44:47,51:54,60)
```

```r
slc_names <- cats(r0)[[1]] |> filter(Value %in% slc) |> pull(Ecosystem_functional_group)
```

```r
efg_freq_tables |>
filter(value %in% slc_names)
```

```r
query_freq_table <- function (qry) 
{
    brks <- set_units(c(0, 1, 5, 10, 15, 100), "%")
    lbls <- c("<1%", "<5%", "5–10%", ">10%", ">15%")
    slc_names <- pull(filter(cats(r0)[[1]], qry), Ecosystem_functional_group)
    tbl <- transmute(filter(efg_freq_tables, value %in% slc_names), 
        `EFG name` = value, `Proportion Pre 1750 (NEAP GET L3 map)` = cut(coalesce(`Proportion Pre 1750`, 
            set_units(0.01, "%")), brks, lbls), `Proportion Extant (NEAP GET L3 map)` = cut(coalesce(`Proportion Extant`, 
            set_units(0.01, "%")), brks, lbls))
    return(tbl)
}
```

# Freshwater

```r
query_freq_table(cats(r0)[[1]]$Realm %in% c("Freshwater","Terrestrial-Freshwater")) |> write.csv("freshwater.csv")
```

# Terrestrial
