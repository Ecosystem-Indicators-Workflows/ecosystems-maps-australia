### Metadata

Metadata download from the ACT Government Geospatial Data Catalogue usin the following url, identifier and parameters:

```python

url='https://www.arcgis.com/sharing/rest/content/items/{uuid}/info/metadata/metadata.xml?format=default'
portal_url = url.format(uuid='abfd3a8744ad4019b139ada2028f2a64')
params = {'format': 'default'}
```

Using the request library to download the file in xml format

```python
r = requests.get(portal_url, params=params)

print("Collection metadata URL: {0}\n".format(r.url))
```

```python
print(r.status_code,r.reason)
```

We use the `xmltodict` library to parse the xml content into a python dictionary, the root entry is the `mdb:MD_Metadata` node:

```python
dict_data = xmltodict.parse(r.content)
metadata = dict_data['mdb:MD_Metadata']
```

We can drill down the different entries in the metadata

```python
for k in metadata.keys():
    if 'mdb' in k:
        print(k)
```

For example the resource lineage information:

```python
display(Markdown(metadata['mdb:resourceLineage']['mrl:LI_Lineage']['mrl:statement']['gco:CharacterString']))
```

Or the citation information:

```python
metadata['mdb:identificationInfo']['mri:MD_DataIdentification']['mri:citation']
```

### Data download

The metadata does not include a download url. 

The data is available at:
https://actmapi-actgov.opendata.arcgis.com/datasets/ACTGOV::actgov-vegetation-map-2023/about

There are several download options, that can be triggered by user clicks.

The API resources allow to download the data in GeoJSON format:

```python
api_url = 'https://services1.arcgis.com/E5n4f1VY84i0xSjy/arcgis/rest/services/ACTGOV_Vegetation_Map_2023/FeatureServer/0/query'
query = {'outFields': '*',
          'where':'1=1',
          'f':'geojson'}

r = requests.get(api_url, params=query, stream=True)
```

Check for HTTP errors

```python
r.raise_for_status()
```

Create folder for the output and path for the downloaded file

```python
output_dir = repodir / 'gisdata' / 'ACT' / 'VegMap_2023'
output_file = output_dir / 'ACTGOV_Vegetation_Map_2023.geojson'
Path(output_dir).mkdir(parents=True, exist_ok=True)
```

Download file from the previous request

```python
with open(output_file, 'wb') as f:
    for chunk in r.iter_content(chunk_size=8192):
        f.write(chunk)
```

Test the downloaded file:

```python
import geopandas as gpd
```
