See publication:

> Olivares-Rojas, J.P., Cook, C.N., Duncan, D.H. and Walsh, J.C. (2025), A Synthesis of Threats Impacting Australian Threatened Ecological Communities. Divers Distrib, 31: e13929. https://doi.org/10.1111/ddi.13929

The data that support the findings of that study are openly available in the online repository (Olivares-Rojas, et al. 2023) https://doi.org/10.26180/24103752.v3. Spatial data of threatened ecological communities can be accessed at (public grids): https://fed.dcceew.gov.au/datasets/25a3f011906c44f0afd287cb647a5d07/about. The data has been generalised to a 100 m grid for the majority of Australia with a 1 km grid used for Western Australia.
