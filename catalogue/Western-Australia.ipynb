> A dataset containing vegetation extent polygons from the mapping of native vegetation in Western Australia. This was originally compiled as part of the vegetation theme of the National Land and Water Resource Audit (NLWRA, 2001). The dataset has been progressively updated by the Department of Agriculture and Food post-NLWRA with assistance of the Department of Environment and Conservation. This has been carried out using digital aerial photography (orthophotos) and satellite imagery acquired 1996 to 2020.
