## Data Download
### Set up

```python
import pyprojroot
import xmltodict
from IPython.display import Markdown, display
import zipfile, requests, io
import os
```

```python
repodir = pyprojroot.find_root(pyprojroot.has_dir(".git"))
```

### Metadata

Metadata download from the DATA VIC portal using the following url, identifier and parameters:

```python
md_url = 'https://metashare.maps.vic.gov.au/geonetwork/srv/api/records/{}/formatters/xml'
# Modelled 1750
ds_id = '853bda8a-a680-5ec4-9cbe-8a33b51eee2b'
# Modelled 2005
ds_id = 'a502df15-7b90-5e96-b1a0-ba29e95558b2'
```

```python
r = requests.get(md_url.format(ds_id))

print("Collection metadata URL: {0}\n".format(r.url))
```

```python
print(r.status_code,r.reason)
```

We use the `xmltodict` library to parse the xml content into a python dictionary, the root entry is the `mdb:MD_Metadata` node:

```python
dict_data = xmltodict.parse(r.content)
metadata = dict_data['mdb:MD_Metadata']
```

```python
for k in metadata.keys():
    if 'mdb' in k:
        print(k)
```

```python
display(Markdown(metadata['mdb:resourceLineage']['mrl:LI_Lineage']['mrl:statement']['gco:CharacterString']))
```

```python
metadata['mdb:identificationInfo']['delwp:MD_DataIdentification']['mri:citation']
```

### Geospatial data

The metadata does not include a download url. 

The data is available through an ordering system (DataShare)

There are several download options, but these need to be triggered by user selection from menus.

The order confirmation includes the order ID, which is valid for a limited period of time.


```python
bucket_url='https://s3.ap-southeast-2.amazonaws.com/cl-isd-prd-datashare-s3-delivery/Order_CAHSK2.zip'
query={'orderid':'A1U1T5'}
r = requests.get(bucket_url, params=query, stream=True)
```

```python
r.raise_for_status()
```

```python
output_dir = repodir / 'gisdata' / 'VIC' / 'EVC_2005'
#output_file = output_dir / 'NV2005_EVCBCS.gdb.zip'
#Path(output_dir).mkdir(parents=True, exist_ok=True)
```

```python
if not os.path.exists(output_dir):
    r = requests.get(bucket_url, params=query,)
    with zipfile.ZipFile(io.BytesIO(r.content)) as zip_ref:
        zip_ref.extractall(output_dir)
    print("Extraction complete.")
else:
    print("Directory already exists. Extraction skipped.")
```
