# NVIS — National Vegetation Information System

The NVIS Information Hierarchy provide six levels of Detailed Vegetation Descriptions. Level 1 provides a general classification of vegetation structure. More structural and species information is added to each level. Level 6 provides the highest detail of vegetation. This includes structure and floristics.

There are over 19,000 distinct vegetation descriptions in the NVIS database, but not all states and territories provide information for the levels 5 and 6.

Generalised NVIS detailed data is used to define 33 Major Vegetation Groups (MVG) and 85 Major Vegetation Subgroups (MVS).

## Data Download

National Vegetation Information System (NVIS) Version 7.0  includes several data products, but here we focus on the extant vectors. 

This download package contains a geodatabase of vector layers delineating the extant ('present') native vegetation types across Australia made available for release under CC BY. The NVIS vegetation attributes contain information on vegetation structure (growth form, height and cover) and floristics (genus and species) as documented in the Australian Vegetation Attribute Manual Version 7.0 (NVIS Technical Working Group, 2017).

download 13G of data and extract to output folder...

```python
unzip $OUTPUT_DIR/NVIS_V7_0_VECTOR_STATE_FILES_EXT_ALL.zip -d $OUTPUT_DIR/
```

Spatial data from the NVIS v7.0 is available in a geodatabase with several components

- *NVIS7_0_AUST_EXT_{State}*:	This dataset is a vector layer delineating the extant ('present') native vegetation types across Australia as a feature class for each State or Territory
- *NVIS7_0_LUT_AUST_DETAIL*	This is a lookup table containing NVIS Version 7.0 vegetation descriptions. The table contains a total of 19,519 NVIS vegetation types.
- *NVIS7_0_LUT_AUST_FLAT*:	This table is a lookup table containing NVIS Version 7.0 vegetation descriptions in a simpler, deconstructed table, allowing for improved analyses and use of the NVIS detailed vegetation descriptions. The table contains a total of 19,519 NVIS vegetation types.
